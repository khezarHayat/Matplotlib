```python
# scatter 3D plot
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

```

```python
batters=pd.read_csv('batter.csv')
```

```python
batters
```

```python
fig=plt.figure()
ax=plt.subplot(projection='3d') # projection parameter is used to specify
ax.scatter3D(batters['runs'],batters['avg'],batters['strike_rate'],marker='+')
ax.set_title('comparision')                                           # other parameter are same to same like other plots
ax.set_xlabel('runs')
```

```python
#3D line plot
```

```python
x=[3,5,0]
y=[6,0,10]
z=[2,8,3]

fig=plt.figure()
ax=plt.subplot(projection='3d')
ax.scatter3D(x,y,z,s=[100,100,100])   # s is used for size of points
ax.plot3D(x,y,z,color='red')
```

```python
# Surface plot
```

```python
x=np.linspace(-20,20,100)
y=np.linspace(-20,20,100)
```

```python
xx,yy=np.meshgrid(x,y)
```

```python
z=xx**2+yy**2
```

```python
z.shape
```

```python
fig=plt.figure()
ax=plt.subplot(projection='3d')
k=ax.plot_surface(xx,yy,z,cmap='viridis')         # want to try diff types of grapgh or diff color combination just used cmap parameter
fig.colorbar(k)                      # it is another fution
```

```python
z=np.sin(xx)+np.cos(yy)               # we used diff math funtion to draw a grapgh 
fig=plt.figure()
ax=plt.subplot(projection='3d')
k=ax.plot_surface(xx,yy,z,cmap='viridis')         # want to try diff types of grapgh or diff color combination just used cmap parameter
fig.colorbar(k)   
```

```python
# Contour plot
```

```python
#if we want to put this 3d grapgh information in 2d we used this grapgh
fig=plt.figure(figsize=(9,6))
ax=plt.subplot()
k=ax.contour(xx,yy,z,cmap='viridis')         # want to try diff types of grapgh or diff color combination just used cmap parameter
fig.colorbar(k)
```

```python
fig=plt.figure(figsize=(9,5))
ax=plt.subplot()
k=ax.contourf(xx,yy,z,cmap='viridis')         # want to try diff types of grapgh or diff color combination just used cmap parameter
fig.colorbar(k)
```
