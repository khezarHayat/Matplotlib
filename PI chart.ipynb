```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%')   # if we want to show percantage on pirchat so we used this parameter
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',colors=['green','blue','black','pink','orange','white']) 
```

```python
plt.pie(df['batsman_runs'],labels=df['batsman'],autopct='%0.1f%%',explode=[0.2,0,0,0,0,0])           #if we want to exract specific value from chat used explode parameter
```
