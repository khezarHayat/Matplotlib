```python
batters=pd.read_csv('batter.csv')
```

```python
batters.head()
```

```python
sample_df=batters.head(25)
```

```python
sample_df
```

```python
plt.figure(figsize=(15,8))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs']) #the size of dot represent each batter runs s funtion is used to increase the size 
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
    #if we want to give labe to each point we used text funtion it is 25 value it is very diffecultto write it one by one used loop
```

```python
#IF WE WANT to find which specific persons are came in specific criteria draw line
plt.figure(figsize=(15,8))
plt.scatter(sample_df['avg'],sample_df['strike_rate'],s=sample_df['runs']) 
plt.axhline(130,color='red') # for strike rate
plt.axvline(30,color='red')  # for acg runs
for i in range(sample_df.shape[0]):
    plt.text(sample_df['avg'].values[i],sample_df['strike_rate'].values[i],sample_df['batter'].values[i])
```

```python
fig,ax=plt.subplots(nrows=2,ncols=1,sharex=True,figsize=(15,6))    # want to display the mutiple grapgh on same data 
                                                        # used shared parameter to shred axis mean each x will be same name
ax[0].scatter(sample_df['avg'],sample_df['strike_rate'],color='red')  # we want to make 2 grapgh so axis ax will be 1,2 and fig for each
ax[1].scatter(sample_df['avg'],sample_df['runs'])              # if want any specific graph just write ax[] and in side braket write number which you want
ax[0].set_title('avg vs striek rate') 
```

```python
fig,ax=plt.subplots(nrows=2,ncols=2,figsize=(10,10)) 
ax[0,0].scatter(sample_df['avg'],sample_df['strike_rate'],color='red')  
ax[0,1].scatter(sample_df['avg'],sample_df['runs'])              # if want any specific graph just write ax[] and in side braket write number which you want
ax[1,0].hist(sample_df['avg'])  
ax[1,1].hist(sample_df['strike_rate'],color='red')                                                # now it is 2d array so if want we will axess like ax[] inside braket we write 2 number ax[0,0],ax[0,1]
```
