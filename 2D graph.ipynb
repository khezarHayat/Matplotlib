```python
plt.plot(batsman['index'],batsman['V Kohli'])
plt.plot(batsman['index'],batsman['RG Sharma'])

plt.title("rohit sharma vs virat kohi caomp")
plt.xlabel("season")                                       #if we want to write the tsxt is each side and graph title
plt.ylabel("runs")
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black')

plt.title("rohit sharma vs virat kohi caomp")
plt.xlabel("season")                                       #want to change the color of lines we also coustamize the color also just knowing the hex code of each color obtain from google
plt.ylabel("runs")
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashed')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashed')

plt.title("rohit sharma vs virat kohi caomp")
plt.xlabel("season")                                       #want to chnge the line style 
plt.ylabel("runs")
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashdot',linewidth=3)
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',linewidth=5)

plt.title("rohit sharma vs virat kohi caomp")
plt.xlabel("season")                                       #want to chnge the line style 
plt.ylabel("runs")
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashdot',linewidth=3,marker='s',markersize=10)
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',linewidth=5,marker='o',markersize=5)

plt.title("rohit sharma vs virat kohi caomp")
plt.xlabel("season")                                       #want to chnge the line style  ,width and ,points where data meet are show
                                                            # for marker we used diff mrker like diamand used'd' for circle 'o' for sqare's'
plt.ylabel("runs")                                           # we also adjust size of marker through markersize parameter
```

```python
plt.plot(batsman['index'],batsman['V Kohli'],color='green',linestyle='dashdot',linewidth=3,marker='s',markersize=10,label='virat')
plt.plot(batsman['index'],batsman['RG Sharma'],color='black',linestyle='dashdot',linewidth=5,marker='o',markersize=5,label='rohit')

plt.title("rohit sharma vs virat kohi caomp")
plt.xlabel("season")                                       # fom the graph we will confiuse which one line is for virat and which one is for rohit so used label parameter
                                                           #and use another funtion called legend and loc parameter is used for to manage the this identification label if we wan we will change from left to right etc
plt.ylabel("runs")  
plt.legend(loc=0)
```
